    📘 3. ['python', 'script.py', './data/sample.txt']

````python
# 正解を出力してください
print('2')
````

---

### ✅ 解説
- 正解：📗 2. `['script.py', './data/sample.txt']`
  - `sys.argv` はコマンドライン引数を格納するリストです。
  - **`sys.argv[0]`** には常に**実行したスクリプト名**が格納されます。
  - **`sys.argv[1]`** 以降には、スクリプト名に続けて渡した**引数**が順番に格納されます。
  - 今回のコマンドでは以下のように格納されます。

| インデックス | 値 | 意味 |
|---|---|---|
| `sys.argv[0]` | `'script.py'` | 実行したスクリプト名 |
| `sys.argv[1]` | `'./data/sample.txt'` | 第一引数（ファイルパス） |

#### ❌ 他の選択肢が誤りである理由
- 📕 1. **`['./data/sample.txt']`** はスクリプト名が含まれていません。`sys.argv[0]` には必ずスクリプト名が格納されるため、引数だけのリストにはなりません
- 📘 3. **`['python', 'script.py', './data/sample.txt']`** は `python` コマンド自体も含まれていますが、`sys.argv` にインタープリタ名（`python`）は格納されません
- 📙 4. **`'./data/sample.txt'`** はリストではなく文字列です。`sys.argv` は必ずリスト形式で値が格納されます

### 🎲 四択問題8 エラー処理の考慮 (3点)

- コマンドライン引数で指定されたファイルパスから、ファイルを読み込むスクリプトがあります。  
- このスクリプトを使用する際に考慮すべき潜在的な問題として、最も適切なものを以下から選べ  

    📕 1. 引数が指定されていない場合  
    📗 2. ファイルが存在しない場合  
    📘 3. ファイルの読み取り権限がない場合    
    📙 4. 上記すべて  
    

````python
# 正解を出力してください
print('4')
````

---

### ✅ 解説
- 正解：📙 4. 上記すべて
  - コマンドライン引数でファイルパスを受け取るスクリプトでは、**複数の潜在的な問題**が発生する可能性があります。
  - これらはいずれも実際の運用で起こりうるエラーであり、すべてに対して適切なエラーハンドリングを行うことが重要です。

```python
import sys

if len(sys.argv) < 2:                      # 📕 引数未指定のチェック
    print("エラー: ファイルパスを指定してください")
    sys.exit(1)

file_path = sys.argv[1]

try:
    with open(file_path, "r", encoding="utf-8") as f:  # 📗📘 存在・権限のチェック
        content = f.read()
except FileNotFoundError:
    print(f"エラー: ファイルが見つかりません: {file_path}")
except PermissionError:
    print(f"エラー: ファイルの読み取り権限がありません: {file_path}")
```

#### 💡 各問題が発生するケースと対処法

- 📕 1. **引数が指定されていない場合**：`sys.argv` にファイルパスが渡されないと `IndexError` が発生します。`len(sys.argv)` で引数の個数を確認することで対処できます
- 📗 2. **ファイルが存在しない場合**：指定したパスにファイルがないと `FileNotFoundError` が発生します。`try/except` で捕捉するか、`os.path.exists()` で事前に確認することで対処できます
- 📘 3. **ファイルの読み取り権限がない場合**：ファイルが存在していても読み取り権限がないと `PermissionError` が発生します。`try/except` で捕捉することで対処できます

---

### ✅ 解説
- 正解：📙 4. `fd.askopenfilename()`
  - `fd.askopenfilename()` は `tkinter.filedialog` モジュールが提供する**ファイル選択ダイアログを開く関数**です。
  - ユーザーがファイルを選択すると、そのファイルパスを文字列として返します。
  - 実際の使用例は以下の通りです。

```python
import tkinter as tk
import tkinter.filedialog as fd

root = tk.Tk()       # ウィンドウの作成
root.withdraw()      # メインウィンドウを非表示にする

file_path = fd.askopenfilename()  # ファイル選択ダイアログを開く
print(file_path)     # 選択されたファイルのパスが表示される
```

#### ❌ 他の選択肢が誤りである理由
- 📕 1. **`root = tk.Tk()`** はtkinterの**メインウィンドウを作成する**コードです。ファイル選択ダイアログを使用する際の前準備として必要ですが、ダイアログを開く関数ではありません
- 📗 2. **`root.withdraw()`** はメインウィンドウを**非表示にする**メソッドです。ファイル選択ダイアログだけを表示させたい場合に使用しますが、ダイアログを開く関数ではありません
- 📘 3. **`fd.withdraw()`** は存在しないメソッドです。`withdraw()` は `tk.Tk()` で生成したウィンドウオブジェクトに対して呼び出すメソッドであり、`filedialog` モジュールには定義されていません
