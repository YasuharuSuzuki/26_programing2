### 🎲 四択問題6 複数グラフのレイアウト (3点)

```python
fig, axs = plt.subplots(2, 3, figsize=(3, 4))
```

- 上記のコードが作成する図全体の 行×列 配置として正しいものはどれか。

    📕 1. 1 行 2 列   
    📗 2. 2 行 3 列   
    📘 3. 3 行 2 列   
    📙 4. 3 行 4 列   

````python
# 正解を出力してください
print('2')
````

---

### ✅ 解説
- 正解：📗 2. 2行3列
  - `plt.subplots(行数, 列数)` の形式で指定するため、`plt.subplots(2, 3)` は**2行3列**のグラフ配置を作成します。
  - `figsize=(3, 4)` は図全体のサイズを **横3インチ・縦4インチ** で指定するものであり、行列の配置とは無関係です。
  - 作成される配置のイメージは以下の通りです。

```
axs[0, 0] | axs[0, 1] | axs[0, 2]   ← 1行目
----------+-----------+-----------
axs[1, 0] | axs[1, 1] | axs[1, 2]   ← 2行目
```

```python
fig, axs = plt.subplots(2, 3, figsize=(3, 4))
# axs は 2行3列の配列として扱える
axs[0, 0].plot([1, 2, 3], [1, 4, 9])  # 1行1列目にグラフを描画
axs[1, 2].bar([1, 2, 3], [3, 6, 9])   # 2行3列目にグラフを描画
```

#### ❌ 他の選択肢が誤りである理由
- 📕 1. **1行2列**は `plt.subplots(1, 2)` と指定した場合の配置です。第一引数の `2` を列数と誤って読んだ場合に混同しやすい選択肢です
- 📘 3. **3行2列**は引数の順序を入れ替えて `plt.subplots(3, 2)` と指定した場合の配置です。行と列を逆に解釈すると混同しやすい選択肢です
- 📙 4. **3行4列**は `figsize=(3, 4)` の値を行列の配置と誤って解釈した場合の選択肢です。`figsize` はあくまで図全体のサイズ（インチ）を指定するものであり、グラフの配置とは無関係です

### 🎲 四択問題7 グラフの目盛線 (3点)

- matplotlibを使用してグラフに目盛線（方眼線）を表示する正しいコードを次から選べ

    📕 1. plt.axis_lines(True)  
    📗 2. plt.grid(True)   
    📘 3. plt.axis_mesh(True)   
    📙 4. plt.ticklines(show=True)   

````python
# 正解を出力してください
print('2')
````

---

### ✅ 解説
- 正解：📗 2. `plt.grid(True)`
  - `plt.grid(True)` はmatplotlibでグラフに**目盛線（グリッド線）を表示する**関数です。
  - 引数に `True` を渡すと目盛線が表示され、`False` を渡すと非表示になります。
  - 目盛線を表示することでグラフの値が読み取りやすくなります。

```python
import matplotlib.pyplot as plt

x = [1, 2, 3, 4, 5]
y = [10, 25, 18, 40, 30]

plt.plot(x, y)
plt.grid(True)   # 目盛線を表示
plt.show()
```

#### ❌ 他の選択肢が誤りである理由
- 📕 1. **`plt.axis_lines(True)`** はmatplotlibに存在しない関数です。`axis` に関連する関数としては `plt.axis()` がありますが、目盛線の表示を制御する機能はありません
- 📘 3. **`plt.axis_mesh(True)`** もmatplotlibに存在しない関数です。`mesh` はメッシュ状のグラフ（`plot_surface()` など）に関連する用語ですが、目盛線の表示とは無関係です
- 📙 4. **`plt.ticklines(show=True)`** もmatplotlibに存在しない関数です。目盛り（tick）に関連する設定は `plt.tick_params()` などで行いますが、目盛線（グリッド線）の表示には `plt.grid()` を使用します

### 🎲 四択問題9 matplotlibのマーカー指定 (3点)

- matplotlibで折れ線グラフを描画する際、円形のマーカーを表示する正しい指定方法を次から選べ。

    📕 1. marker='●'  
    📗 2. marker='circle'  
    📘 3. marker='round'  
    📙 4. marker='o'   

````python
# 正解を出力してください
print('4')
````

---

### ✅ 解説
- 正解：📙 4. `marker='o'`
  - matplotlibでは `plt.plot()` の `marker` 引数に **`'o'`（小文字のアルファベット）** を指定することで円形のマーカーを表示できます。
  - マーカーは各データ点の位置に描画され、データの位置を視覚的にわかりやすくする役割を持ちます。

```python
import matplotlib.pyplot as plt

x = [1, 2, 3, 4, 5]
y = [10, 25, 18, 40, 30]

plt.plot(x, y, marker='o')  # 円形マーカーを表示
plt.show()
```

#### 💡 主なマーカー指定の一覧
| 指定値 | マーカーの形 |
|--------|------------|
| `'o'`  | 円形 |
| `'s'`  | 正方形 |
| `'^'`  | 上向き三角形 |
| `'*'`  | 星形 |
| `'+'`  | プラス記号 |

#### ❌ 他の選択肢が誤りである理由
- 📕 1. **`marker='●'`** は実際の円形文字を指定していますが、matplotlibのマーカー指定はアルファベットや記号による**決められた文字列**で行う必要があり、Unicode文字は使用できません
- 📗 2. **`marker='circle'`** は直感的でわかりやすい指定に見えますが、matplotlibでは英単語によるマーカー指定はサポートされていません
- 📘 3. **`marker='round'`** も同様に英単語によるマーカー指定はサポートされておらず、このような指定をしても円形マーカーは表示されません
